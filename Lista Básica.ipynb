```python
print ("Dobro e Triplo")

a = float (input ("Insira um número: "))

print ("O dobro: ", a * 2, "\nO triplo: ", a * 3)
```

```python
print ("Antecessor e Sucessor")

a = float (input ("Insira um número: "))

print ("O antecessor é: ", a - 1, "\nO sucessor é: ", a + 1)
```

```python
print ("Área do triângulo")

h = float (input ("Insira a altura: "))
b = float (input ("Insira a base: "))

print ("A área do triângulo é: ", (h + b) / 2)
```

```python
print ("Perimetro do retângulo")

lado1 = float (input ("Insira o lado 1: "))
lado2 = float (input ("Insira o lado 2: "))
ladoA = float (input ("Insira o lado A: "))
ladoB = float (input ("Insira o lado B: "))

print ("O perimetro do retângulo é: ", lado1 + lado2 + ladoA + ladoB)
```
